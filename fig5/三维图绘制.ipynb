```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D
from scipy.interpolate import griddata

import pandas as pd

import os
```

```python
xy = pd.read_csv('D:/fsh/旧/data/paad.tissue_positions_list.csv', index_col=0)
data = pd.read_csv('D:/fsh/旧/data/paaddealT2.csv')
```

```python
# 合并 xy 和 z 数据
xyz = pd.merge(data, xy, left_on='Unnamed: 0', right_index=True, how='inner')
```

```python
xyz.
```

```python
# 提取相关数据
x = xyz['4217'].values
y = xyz['4669'].values
z = xyz['VCAN'].values

# 创建 2D 网格坐标
xi = np.linspace(-x.max(), -x.min(), 100)
yi = np.linspace(y.max(), y.min(), 100)
Xi, Yi = np.meshgrid(xi, yi)

# 使用 griddata 进行插值
Zi = griddata((-x, y), z, (Xi, Yi), method='linear')

# 创建 3D 图形
fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')

# 绘制 3D 表面图
surf = ax.plot_surface(Xi, Yi, Zi, rstride=1, cstride=1, cmap='inferno', alpha=0.75)

# 设置坐标轴标签和标题
ax.set_xlabel('X')
ax.set_ylabel('Y')
ax.set_zlabel('Z')
ax.set_title('3D Surface Plot')

# 设置 70 仰角和 0 方位角
ax.view_init(elev=70, azim=0)

# 设置坐标轴颜色为透明
ax.xaxis.set_pane_color((1.0, 1.0, 1.0, 0.0))
ax.yaxis.set_pane_color((1.0, 1.0, 1.0, 0.0))
ax.zaxis.set_pane_color((1.0, 1.0, 1.0, 0.0))

# 设置坐标轴线条颜色为透明
ax.xaxis._axinfo["grid"]['color'] = (1,1,1,0)
ax.yaxis._axinfo["grid"]['color'] = (1,1,1,0)
ax.zaxis._axinfo["grid"]['color'] = (1,1,1,0)

# 添加图例
cbar = fig.colorbar(surf, shrink=0.5, aspect=10)


# 保存图片
plt.savefig(os.path.join("D:/fsh/xff/paad_VCAN.pdf"))
plt.close()
```
